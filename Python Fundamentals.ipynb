A comparação entre o uso de uma variável temporária e a sintaxe
`a, b = b, a` demonstra a maturidade do código. A abordagem *Pythonic*
não é apenas mais curta; ela reduz a verbosidade e minimiza as chances
de erro na manipulação do estado, algo fundamental em algoritmos
complexos de ordenação.

## 4. Utilizando o Poder das Bibliotecas Standard (`math`, `random`, `calendar`)

### Implementações com Bibliotecas — Programas 5, 8, 9 e 30

``` python
import random
import calendar
import math

# Programa 5: Geração de número aleatório
# Útil para simulações e testes automatizados
print(f"Número aleatório entre 1 e 100: {random.randint(1, 100)}")

# Programa 8: Exibição de calendário
ano = int(input("Ano: "))
mes = int(input("Mês: "))
print(calendar.month(ano, mes))

# Programa 9: Equação Quadrática (ax^2 + bx + c = 0)
a = float(input("Coeficiente a: "))
b = float(input("Coeficiente b: "))
c = float(input("Coeficiente c: "))

# O discriminante determina a natureza das raízes
delta = b**2 - 4*a*c

if delta > 0:
    raiz1 = (-b + math.sqrt(delta)) / (2*a)
    raiz2 = (-b - math.sqrt(delta)) / (2*a)
    print(f"Duas raízes reais: {raiz1} e {raiz2}")
elif delta == 0:
    raiz = -b / (2*a)
    print(f"Raiz real única: {raiz}")
else:
    real = -b / (2*a)
    imag = math.sqrt(abs(delta)) / (2*a)
    print(f"Raízes complexas: {real} + {imag}i e {real} - {imag}i")

# Programa 30: Logaritmo Natural
num_log = float(input("Digite um número para calcular ln(n): "))
if num_log > 0:
    print(f"O logaritmo natural de {num_log} é: {math.log(num_log)}")
```

A utilização do módulo `math` garante a precisão necessária para
cálculos científicos, enquanto o `calendar` fornece utilitários prontos
que economizam tempo de desenvolvimento em aplicações que dependem de
datas e formatação de tempo.

## 5. Lógica Decisória e Fluxo de Controle

Estruturas condicionais (`if`/`elif`/`else`) são o “cérebro” de um
programa. Elas permitem que o software responda dinamicamente a
diferentes cenários, criando caminhos lógicos que tratam exceções e
regras específicas com precisão.

### Decisões Inteligentes — Programas 11 a 13

``` python
# Programa 11: Positivo, Negativo ou Zero
num = float(input("Digite um número para análise: "))
if num > 0:
    print("O número é Positivo")
elif num == 0:
    print("O número é Zero")
else:
    print("O número é Negativo")

# Programa 12: Par ou Ímpar
# Uso do operador módulo (%) para verificar divisibilidade
num_check = int(input("Digite um número inteiro: "))
if num_check % 2 == 0:
    print("Este número é Par")
else:
    print("Este número é Ímpar")

# Programa 13: Ano Bissexto
# Demonstração de condições múltiplas encadeadas
ano_b = int(input("Digite o ano para verificar se é bissexto: "))
if (ano_b % 400 == 0) and (ano_b % 100 == 0):
    print(f"{ano_b} é um ano bissexto")
elif (ano_b % 4 == 0) and (ano_b % 100 != 0):
    print(f"{ano_b} é um ano bissexto")
else:
    print(f"{ano_b} não é um ano bissexto")
```

A identificação de anos bissextos é um exemplo clássico de como regras
de negócio complexas exigem precisão algorítmica. Sem essa lógica,
sistemas de agendamento e calendários falhariam em manter a integridade
temporal.

## 6. Algoritmos Iterativos, Primos e Sequências Numéricas

Processar conjuntos de dados e identificar padrões matemáticos exige o

### Iteração e Recursão — Programas 14 a 18, 27 e 28

``` python
# Programa 14: Verificar Número Primo (Abordagem Pythonic)
num_primo = int(input("Verificar se é primo: "))
if num_primo > 1:
    for i in range(2, num_primo):
        if (num_primo % i) == 0:
            print(f"{num_primo} não é primo")
            break
    else:
        # O bloco 'else' do loop 'for' executa se o loop não for interrompido pelo 'break'
        print(f"{num_primo} é um número primo")
else:
    print(f"{num_primo} não é primo")

# Programa 15: Primos em um Intervalo (1 a 10)
print("Primos entre 1 e 10:")
for n in range(1, 11):
    if n > 1:
        for j in range(2, n):
            if (n % j) == 0:
                break
        else:
            print(n, end=" ")
print("\n")

# Programa 17: Tabuada de Multiplicação
tab = int(input("Tabuada de qual número? "))
for i in range(1, 11):
    print(f"{tab} X {i} = {tab * i}")

# Programa 18: Fibonacci Iterativo (Performance O(n))
n_terms = int(input("Quantos termos da sequência de Fibonacci? "))
n1, n2 = 0, 1
count = 0
print("Sequência Fibonacci (Iterativa):")
while count < n_terms:
    print(n1)
    nth = n1 + n2
    n1 = n2
    n2 = nth
    count += 1

# Programa 27: Fibonacci Recursivo (Complexidade O(2^n))
def recur_fibo(n):
    if n <= 1:
        return n
    return recur_fibo(n-1) + recur_fibo(n-2)

# Programa 16: Fatorial Iterativo
num_f = int(input("Fatorial de: "))
fatorial = 1
for i in range(1, num_f + 1):
    fatorial *= i
print(f"Fatorial iterativo de {num_f}: {fatorial}")

# Programa 28: Fatorial Recursivo
def recur_factorial(n):
    return 1 if (n == 1 or n == 0) else n * recur_factorial(n-1)
```

Comparar abordagens iterativas e recursivas é vital. Enquanto o
Fibonacci recursivo é matematicamente elegante, sua complexidade
exponencial, $O(2^n)$, o torna inviável para valores altos. A versão
iterativa, $O(n)$, demonstra uma maturidade técnica necessária para
sistemas de alta performance.

## 7. Verificação de Propriedades Numéricas (Armstrong, Disarium, Happy)

Explorar propriedades numéricas especiais desenvolve o pensamento
analítico. Estes exercícios exigem a decomposição de números em dígitos,
uma técnica essencial em tarefas de limpeza e processamento de dados
(*data cleaning*).

### Propriedades de Dígitos — Programas 19, 20, 43, 44, 46 e 49

``` python
# Programa 19/20: Número de Armstrong
num_arm = int(input("Verificar Número de Armstrong: "))
num_str = str(num_arm)
ordem = len(num_str)
soma_arm = sum(int(digito) ** ordem for digito in num_str)
if soma_arm == num_arm:
    print(f"{num_arm} é um número de Armstrong")

# Programa 43/44: Número de Disarium (Busca em Intervalo)
def is_disarium(n):
    # Soma de cada dígito elevado à sua posição (base 1)
    return n == sum(int(d) ** (i + 1) for i, d in enumerate(str(n)))

print("Números Disarium entre 1 e 100:")
for i in range(1, 101):
    if is_disarium(i):
        print(i, end=" | ")
print("\n")

# Programa 46: Happy Numbers
def is_happy(n):
    vistos = set()
    while n != 1 and n not in vistos:
        vistos.add(n)
        n = sum(int(d)**2 for d in str(n))
    return n == 1

h_num = int(input("Verificar se o número é 'Feliz' (Happy Number): "))
print(f"É feliz? {is_happy(h_num)}")

# Programa 49: Número de Harshad
num_h = int(input("Número de Harshad: "))
soma_digitos = sum(int(d) for d in str(num_h))
if num_h % soma_digitos == 0:
    print(f"{num_h} é um número de Harshad (divisível pela soma de seus dígitos)")
```

Tratar números como strings permite extrair padrões que não são óbvios
apenas através de aritmética tradicional. Essa versatilidade é
fundamental para validar formatos de documentos (como CPF/CNPJ) e
sanitizar entradas de usuários.

## 8. Estruturas de Dados: Manipulação de Arrays (Listas)

Listas em Python são estruturas dinâmicas fundamentais. Elas permitem
armazenar e processar coleções de objetos de maneira flexível, servindo
como base para qualquer manipulação de dados em larga escala.

### Gerenciamento de Listas — Programas 32 a 35

``` python
# Programa 32/33: Soma e Maior Elemento
meu_array = [10, 20, 30, 99]

# Lógica Manual (Fundamental para entender algoritmos)
soma_manual = 0
for x in meu_array:
    soma_manual += x

# Lógica Pythonic (Uso de built-ins implementados em C para performance)
soma_otimizada = sum(meu_array)
maior = max(meu_array)
print(f"Soma: {soma_otimizada}, Maior: {maior}")

# Programa 34: Rotação de Array
def rotate_array(arr, d):
    # Abordagem por fatiamento (slicing) - mais eficiente e produtiva
    return arr[d:] + arr[:d]

arr_rot = [1, 2, 3, 4, 5]
print(f"Array rotacionado em 2 posições: {rotate_array(arr_rot, 2)}")

# Programa 35: Verificar se um Array é Monotônico
def is_monotonic(arr):
    # Um array é monotônico se for inteiramente crescente ou decrescente
    crescente = decrescente = True
    for i in range(1, len(arr)):
        if arr[i] > arr[i-1]:
            decrescente = False
        if arr[i] < arr[i-1]:
            crescente = False
    return crescente or decrescente
```

O uso de funções embutidas como `sum()` e `max()` é preferível em
produção, pois são implementadas em C (CPython), oferecendo performance
superior. Contudo, dominar a implementação manual e técnicas de
fatiamento (*slicing*) é um diferencial em desafios de codificação.

## 9. Ferramentas Práticas e Utilitários de Saúde

A culminação de toda a lógica anterior é a criação de ferramentas que
resolvem problemas reais. Transformar fórmulas complexas em utilitários
funcionais é a habilidade mais valorizada no mercado.

### Aplicações Reais — Programas 26 e 29

``` python
# Programa 26: Calculadora Simples
def calculadora():
    print("1.Soma | 2.Subtração | 3.Multiplicação | 4.Divisão")
    escolha = input("Escolha a operação (1-4): ")
    n1 = float(input("Primeiro número: "))
    n2 = float(input("Segundo número: "))
    
    if escolha == '1': print(f"Resultado: {n1 + n2}")
    elif escolha == '2': print(f"Resultado: {n1 - n2}")
    elif escolha == '3': print(f"Resultado: {n1 * n2}")
    elif escolha == '4': 
        if n2 != 0: print(f"Resultado: {n1 / n2}")
        else: print("Erro: Divisor zero.")

# Programa 29: Calculador de IMC (Índice de Massa Corporal)
altura = float(input("Sua altura em metros (ex: 1.75): "))
peso = float(input("Seu peso em kg (ex: 70): "))
imc = round(peso / (altura ** 2), 2)
print(f"Seu IMC é: {imc}")

# Classificação baseada em padrões de saúde
if imc <= 18.5:
    print("Classificação: Abaixo do peso")
elif imc <= 24.9:
    print("Classificação: Peso normal")
elif imc <= 29.9:
    print("Classificação: Sobrepeso")
else:
    print("Classificação: Obeso")
```

A capacidade de traduzir uma necessidade de saúde ou de negócio em um
